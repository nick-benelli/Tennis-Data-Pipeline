```python
import ssl
import urllib.request

import truststore

# use macOS Keychain (needed behind corporate TLS-inspecting proxies like Zscaler);
# also force a fresh opener since the kernel may have already built one with the old context
truststore.inject_into_ssl()
urllib.request.install_opener(
    urllib.request.build_opener(urllib.request.HTTPSHandler(context=truststore.SSLContext(ssl.PROTOCOL_TLS_CLIENT)))
)

import pandas as pd
```

```python
df = pd.read_csv("https://stats.tennismylife.org/data/2026.csv")

df
```

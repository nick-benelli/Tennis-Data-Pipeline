```python
## Aggregate to Tournament Level

One row per `source_event_key`, so tournaments can be compared across datasets for a given year.
```

```python
# Knockout rounds, ordered from earliest to the final.
KNOCKOUT_ROUND_ORDER = ["R128", "R64", "R32", "R16", "QF", "SF", "F"]


def _draw_size(rounds: pd.Series) -> object:
    """Draw size, inferred from how many knockout rounds were played.

    Tennis-Data UK always labels a tournament's opening round "R128"
    regardless of its actual size (e.g. a 32-draw ATP 250's first round is
    still labelled "R128"), so the round name itself isn't reliable - only
    the *number* of distinct knockout rounds played is: each round halves
    the field, so draw_size = 2 ** (number of knockout rounds).
    """
    rounds_present = set(rounds.dropna().astype(str))
    if "RR" in rounds_present:
        # Season-ending round-robin events (e.g. the ATP/WTA Finals) are 8-player.
        return 8
    knockout_rounds = rounds_present & set(KNOCKOUT_ROUND_ORDER)
    return 2 ** len(knockout_rounds) if knockout_rounds else pd.NA


def _final_winner(group: pd.DataFrame) -> object:
    final = group.loc[group["round"] == "F", "winner_name"]
    return final.iloc[0] if not final.empty else pd.NA


tournaments_uk = (
    df_uk.groupby("source_event_key", sort=False)
    .agg(
        source=("source", "first"),
        tour=("tour", "first"),
        year=("year", "first"),
        uk_tournament_id=(
            "uk_tournament_id",
            lambda s: sorted(s.dropna().unique().tolist()),
        ),
        tournament_name=("tournament_name", "first"),
        location=("location", "first"),
        start_date=("match_date", "min"),
        end_date=("match_date", "max"),
        series=("series", "first"),
        is_outdoor=("is_outdoor", "first"),
        surface=("surface", "first"),
        draw_size=("round", _draw_size),
        best_of=("best_of", "first"),
        num_matches=("round", "count"),
    )
    .reset_index()
)

winner_uk = (
    df_uk.groupby("source_event_key", sort=False)
    .apply(_final_winner, include_groups=False)
    .rename("winner_name")
)

tournaments_uk = tournaments_uk.merge(winner_uk, on="source_event_key", how="left")

tournaments_uk
```

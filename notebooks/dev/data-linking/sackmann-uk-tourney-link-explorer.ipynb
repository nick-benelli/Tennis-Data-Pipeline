## Programmatic Tournament Matching

Sizes differ (UK: 60, Sackmann: 63) so this isn't a 1:1 join. We match on
normalized tournament name similarity + start-date proximity, then greedily
assign the best-scoring unique pairs. Anything left over is kept with a
blank id on the other side so the result can be hand-reviewed/edited.

Dedup/matching below is keyed on `source_event_key`, not `uk_tournament_id`
- the latter resets 1-60 every year, so it's only safe to use as a unique
key within a single (tour, year) like this notebook does; `source_event_key`
is globally unique and future-proofs this if the notebook ever expands to
multiple years at once. `location`/`uk_name`/`sackmann_name` are carried
alongside purely for manual spot-checking - the target end state is a plain
3-column crosswalk: `year`, `uk_tournament_id`, `sackmann_tourney_id`.

```python
import re
import unicodedata
from difflib import SequenceMatcher


def normalize_tournament_name(name: str) -> str:
    """Lowercase, strip accents/punctuation so e.g. "Queen's Club" == "Queens Club"."""
    name = unicodedata.normalize("NFKD", str(name)).encode("ascii", "ignore").decode("ascii")
    name = name.lower()
    name = re.sub(r"[^a-z0-9]+", " ", name)
    return name.strip()


# UK names are often sponsor names (e.g. "BMW Open") while Sackmann uses the host
# city (e.g. "Munich"), so we keep the UK "location" around too and match against both.
uk = df_uk_tourneys[["uk_tournament_id", "source_event_key", "tournament_name", "location", "surface", "start_date", "end_date"]].copy()
sack = df_sackman_tourneys[["tourney_id", "tournament_name", "surface", "start_date", "end_date"]].copy()

uk["start_date"] = pd.to_datetime(uk["start_date"])
sack["start_date"] = pd.to_datetime(sack["start_date"])

uk["name_norm"] = uk["tournament_name"].map(normalize_tournament_name)
uk["location_norm"] = uk["location"].map(normalize_tournament_name)
sack["name_norm"] = sack["tournament_name"].map(normalize_tournament_name)
```

```python
MAX_DATE_DIFF_DAYS = 10  # same tournament shouldn't drift further than this between sources


def date_score(uk_start, sack_start, max_days: int = MAX_DATE_DIFF_DAYS) -> float:
    """
    Score based on how close the start dates are.

    Args:
        uk_start (datetime): Start date of the UK tournament.
        sack_start (datetime): Start date of the Sackmann tournament.
        max_days (int): Maximum allowed difference in days for scoring.

    Returns:
        float: A score between 0 and 1 based on the closeness of the start dates.
    """
    diff_days = abs((uk_start - sack_start).days)
    return max(0.0, 1 - diff_days / max_days)


candidates = []
for uk_row in uk.itertuples():
    for sack_row in sack.itertuples():
        # try the UK sponsor name AND the UK host-city location against the Sackmann name
        name_score = max(
            SequenceMatcher(None, uk_row.name_norm, sack_row.name_norm).ratio(),
            SequenceMatcher(None, uk_row.location_norm, sack_row.name_norm).ratio(),
        )
        d_score = date_score(uk_row.start_date, sack_row.start_date)
        surface_match = float(uk_row.surface == sack_row.surface)
        score = 0.5 * name_score + 0.4 * d_score + 0.1 * surface_match
        candidates.append(
            {
                "uk_tournament_id": uk_row.uk_tournament_id,
                "uk_source_event_key" : uk_row.source_event_key,
                "sackmann_tourney_id": sack_row.tourney_id,
                "location": uk_row.location,
                "uk_name": uk_row.tournament_name,
                "sackmann_name": sack_row.tournament_name,
                "name_score": name_score,
                "date_score": d_score,
                "surface_match": surface_match,
                "score": score,
            }
        )

candidates_df = pd.DataFrame(candidates).sort_values("score", ascending=False)
candidates_df.head()
```

```python
MIN_MATCH_SCORE = 0.5  # candidates below this are treated as "no match"

# uk_tournament_id resets 1-60 every year, so source_event_key (globally unique -
# see handler/uk/cleaner/common.add_source_event_key) is the key used to dedupe here.
matched_uk_keys: set = set()
matched_sack_ids: set = set()
matches = []

for row in candidates_df.itertuples():
    if row.score < MIN_MATCH_SCORE:
        break  # sorted descending, nothing better remains
    if row.uk_source_event_key in matched_uk_keys or row.sackmann_tourney_id in matched_sack_ids:
        continue
    matched_uk_keys.add(row.uk_source_event_key)
    matched_sack_ids.add(row.sackmann_tourney_id)
    matches.append(
        {
            "year": YEAR,
            "uk_tournament_id": row.uk_tournament_id,
            "uk_source_event_key": row.uk_source_event_key,
            "sackmann_tourney_id": row.sackmann_tourney_id,
            "score": row.score,
            "location": row.location,
            "uk_name": row.uk_name,
            "sackmann_name": row.sackmann_name,
        }
    )

matched_df = pd.DataFrame(matches)
print(f"Matched {len(matched_df)} / {len(uk)} UK tourneys and {len(matched_df)} / {len(sack)} Sackmann tourneys")
matched_df.sort_values("score")[["score", "location", "uk_name", "sackmann_name"]].head(15)  # weakest first
```

```python
# Manual verification pass: eyeball every match's location/name pairing before trusting it.
matched_df[["score", "location", "uk_name", "sackmann_name"]].sort_values("score")
```

```python
unmatched_uk = uk.loc[~uk["source_event_key"].isin(matched_uk_keys), ["uk_tournament_id", "source_event_key"]]
unmatched_sack = sack.loc[~sack["tourney_id"].isin(matched_sack_ids), "tourney_id"]

unmatched_uk_df = pd.DataFrame(
    {
        "year": YEAR,
        "uk_tournament_id": unmatched_uk["uk_tournament_id"],
        "uk_source_event_key": unmatched_uk["source_event_key"],
        "sackmann_tourney_id": pd.NA,
    }
)
unmatched_sack_df = pd.DataFrame(
    {"year": YEAR, "uk_tournament_id": pd.NA, "uk_source_event_key": pd.NA, "sackmann_tourney_id": unmatched_sack}
)

# uk_source_event_key travels alongside for now (it's what guarantees uniqueness here) -
# the eventual persisted crosswalk is just year/uk_tournament_id/sackmann_tourney_id.
link_df = pd.concat(
    [matched_df[["year", "uk_tournament_id", "uk_source_event_key", "sackmann_tourney_id"]], unmatched_uk_df, unmatched_sack_df],
    ignore_index=True,
)
link_df = link_df.sort_values(["uk_source_event_key", "sackmann_tourney_id"], na_position="last").reset_index(drop=True)
link_df
```

```python
# Hand-edit this file to correct any bad matches/fill in the blanks, then re-load it as the source of truth.
link_df.to_csv(f"uk_sackmann_tourney_link_{TOUR.lower()}_{YEAR}.csv", index=False)
```

## Bonus: Sackmann's `tourney_number` *is* the permanent ATP tournament id

Cross-checking against the ATP site's own tournament ids (Montpellier=375,
Delray Beach=499, ...) shows Sackmann's `tourney_number` is that exact same
permanent id, not a Sackmann-only invention - e.g. `2025-0375` in Sackmann
is Montpellier, `2025-0499` is Delray Beach. That means once a UK tournament
is linked to a Sackmann `tourney_id`, we get the permanent ATP id for free,
with no fuzzy matching required on that leg.

Two caveats found by checking every non-Davis-Cup row across the whole
history (not just 2025): one Olympics row uses an `O16`-style code (not a
real ATP tour id), and **2016-2020** use an unrelated `M0xx` numbering scheme
for ~13-14 Masters/500-level tournaments each of those years (verified those
years have no separate numeric-id duplicate row for the same event - it's a
genuine gap in this range, not visible in other years). Both are non-numeric
so they parse to a blank id below rather than a wrong one.

```python
df_sackmann_all_years = pd.read_csv(
    PROJECT_DIR / "data/clean/sackmann" / TOUR / f"tournaments/sackmann_{TOUR.lower()}_tournaments.csv",
    dtype={"tourney_number": str},
)
df_sackmann_all_years = df_sackmann_all_years.loc[df_sackmann_all_years["tourney_level"] != "D"]

is_numeric_id = df_sackmann_all_years["tourney_number"].str.match(r"^\d+$")
print(f"{is_numeric_id.sum()}/{len(df_sackmann_all_years)} rows have a numeric tourney_number")
df_sackmann_all_years.loc[~is_numeric_id, ["year", "tourney_number", "tourney_level", "tournament_name"]]
```

```python
def extract_atp_tournament_id(sackmann_tourney_id):
    if pd.isna(sackmann_tourney_id):
        return pd.NA
    number = str(sackmann_tourney_id).split("-", 1)[1]
    return int(number) if number.isdigit() else pd.NA


link_df["atp_tournament_id"] = link_df["sackmann_tourney_id"].map(extract_atp_tournament_id).astype("Int64")

# spot-check the user's discovery: Montpellier -> 375, Delray Beach -> 499
check = link_df.merge(
    uk[["source_event_key", "location"]], left_on="uk_source_event_key", right_on="source_event_key", how="left"
)
check.loc[check["location"].isin(["Montpellier", "Delray Beach"])]
```

## Build a `location_key -> atp_tournament_id` crosswalk

The fuzzy name/date/surface matching above is only needed because we don't
have a permanent id to key on directly. But a UK tournament's `location`
(host city) is stable across years, so once a location has been matched here
it never needs to be fuzzy-matched again - just normalize the location and
look it up.

One wrinkle: a location isn't always unique to one tournament - Paris hosts
both Roland Garros (520) and the Paris Masters (352) - so a bare `location`
key is ambiguous for those. The code below detects that automatically and
falls back to `location + tournament_name` for any ambiguous location only.

This becomes a growable, mostly-static reference table: rerun this notebook
for another year, `pd.concat` + `drop_duplicates` the result onto the
existing crosswalk CSV, and only genuinely new locations need any
matching/review at all.

```python
resolved = link_df.dropna(subset=["uk_tournament_id", "atp_tournament_id"]).merge(
    uk[["source_event_key", "location", "location_norm", "name_norm"]],
    left_on="uk_source_event_key",
    right_on="source_event_key",
    how="left",
)

# a location mapping to more than one id (e.g. "Paris" hosts both Roland Garros and the Paris
# Masters) isn't safe to key on alone - fall back to location+tournament name for those.
ids_per_location = resolved.groupby("location_norm")["atp_tournament_id"].nunique()
ambiguous_locations = set(ids_per_location[ids_per_location > 1].index)
print(f"Ambiguous location-only keys: {sorted(ambiguous_locations)}")

resolved["location_key"] = resolved["location_norm"].where(
    ~resolved["location_norm"].isin(ambiguous_locations),
    resolved["location_norm"] + "|" + resolved["name_norm"],
)

location_crosswalk = (
    resolved[["location_key", "atp_tournament_id", "location"]]
    .drop_duplicates(subset="location_key")
    .sort_values("location_key")
    .reset_index(drop=True)
)

print(f"{len(location_crosswalk)} location keys resolved this run")
location_crosswalk
```

```python
CROSSWALK_PATH = Path("atp_location_id_crosswalk.csv")

if CROSSWALK_PATH.exists():
    existing_crosswalk = pd.read_csv(CROSSWALK_PATH)
    location_crosswalk = (
        pd.concat([existing_crosswalk, location_crosswalk], ignore_index=True)
        .drop_duplicates(subset="location_key", keep="first")
        .sort_values("location_key")
        .reset_index(drop=True)
    )

location_crosswalk.to_csv(CROSSWALK_PATH, index=False)
print(f"{len(location_crosswalk)} total location keys saved to {CROSSWALK_PATH}")
```

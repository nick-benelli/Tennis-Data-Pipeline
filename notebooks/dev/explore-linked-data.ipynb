## Feature audit: which columns to keep for EDA/modelling

Sackmann is the gold-standard source; `tennis-data-uk` mainly adds value where
Sackmann has nothing (betting odds, `series`, `location`) or where linkage QA
lives (`canonical_*`, `match_method`, `review_flag`, `data_quality_flag`).
Everything else duplicated between the two sources should default to the
Sackmann version.

Cells below profile every merged column (dtype, null %, n unique) grouped by
where it came from, so we can decide keep vs. drop with real numbers instead
of guessing.

```python
sack_cols = set(df_sack.columns)
uk_cols = set(df_uk_enriched.columns)

# columns that existed in both frames pre-merge got a _sack/_uk suffix;
# recover their shared "concept" name so we can compare side by side
shared_concepts = sorted(sack_cols & uk_cols)
sack_only = sorted(sack_cols - uk_cols)
uk_only = sorted(uk_cols - sack_cols)

print(f"Shared column names (suffixed _sack/_uk in df_merge): {len(shared_concepts)}")
print(shared_concepts)
print(f"\nSackmann-only columns: {len(sack_only)}")
print(sack_only)
print(f"\nUK-only columns: {len(uk_only)}")
print(uk_only)
```

```python
audit = pd.DataFrame({
    "dtype": df_merge.dtypes.astype(str),
    "null_pct": (df_merge.isna().mean() * 100).round(1),
    "n_unique": df_merge.nunique(dropna=True),
})
audit["origin"] = [
    "shared" if c.rsplit("_", 1)[-1] in ("sack", "uk") and c.rsplit("_", 1)[0] in shared_concepts
    else ("sackmann" if c in sack_only else ("uk" if c in uk_only else "join_key"))
    for c in audit.index
]
with pd.option_context("display.max_rows", None):
    display(audit.sort_values(["origin", "null_pct"]))
```

```python
# how much do the two sources actually agree on the fields they both record?
pairs = [
    ("surface_sack", "surface_uk"),
    ("best_of_sack", "best_of_uk"),
    ("round_sack", "round_uk"),
    ("round_sack", "canonical_round"),
    ("winner_rank_sack", "winner_rank_uk"),
    ("loser_rank_sack", "loser_rank_uk"),
    ("winner_rank_points_sack", "winner_rank_points_uk"),
    ("loser_rank_points_sack", "loser_rank_points_uk"),
    ("winner_id", "canonical_winner_id"),
    ("loser_id", "canonical_loser_id"),
]
for a, b in pairs:
    left = df_merge[a].astype(str)
    right = df_merge[b].astype(str)
    both_present = df_merge[a].notna() & df_merge[b].notna()
    agree = (left[both_present] == right[both_present]).mean() * 100
    print(f"{a:28s} vs {b:28s}: agree={agree:5.1f}%  (n={both_present.sum()})")

print("\nseries value counts:\n", df_merge["series"].value_counts())
print("\nmatch_method value counts:\n", df_merge["match_method"].value_counts())
print("\nreview_flag value counts:\n", df_merge["review_flag"].value_counts())
print("\ndata_quality_flag value counts:\n", df_merge["data_quality_flag"].value_counts(dropna=False))
```

```python
# surface agreement came back 0% - inspect raw values to see if it's just a casing/label difference
print(df_merge["surface_sack"].value_counts())
print(df_merge["surface_uk"].value_counts())
print(df_merge[["surface_sack", "surface_uk"]].drop_duplicates())
```

```python
# WTA rank_points agreement came back 0% (vs ATP's ~98%) - inspect why
cols = ["winner_rank_points_sack", "winner_rank_points_uk", "loser_rank_points_sack", "loser_rank_points_uk"]
display(df_merge[cols].dropna().head(10))
print(df_merge[cols].dtypes)
diff = df_merge["winner_rank_points_sack"] - df_merge["winner_rank_points_uk"]
print("\nwinner_rank_points_sack - winner_rank_points_uk, describe():")
print(diff.describe())
```

```python
# fix: previous check used astype(str), which breaks for Int64 vs float64
# ("3925" != "3925.0"). Redo numerically for the numeric pairs.
numeric_pairs = [
    ("winner_rank_sack", "winner_rank_uk"),
    ("loser_rank_sack", "loser_rank_uk"),
    ("winner_rank_points_sack", "winner_rank_points_uk"),
    ("loser_rank_points_sack", "loser_rank_points_uk"),
]
for a, b in numeric_pairs:
    left = df_merge[a].astype("Float64")
    right = df_merge[b].astype("Float64")
    both_present = left.notna() & right.notna()
    agree = (left[both_present] == right[both_present]).mean() * 100
    print(f"{a:28s} vs {b:28s}: agree={agree:5.1f}%  (n={both_present.sum()})")

# surface case-difference is a real (non-bug) 0%: confirm same pattern as ATP
print()
print(df_merge[["surface_sack", "surface_uk"]].drop_duplicates())
```

```python
# surface_sack="Hard" vs surface_uk="clay" is a genuine mismatch, not casing - find it
mismatch = df_merge[df_merge["surface_sack"].str.lower() != df_merge["surface_uk"].str.lower()]
display(mismatch[[
    "canonical_match_key", "tourney_name", "tourney_date", "location",
    "surface_sack", "surface_uk", "round_sack", "winner_name_sack", "loser_name_sack",
]])
```

## Findings & recommended feature set

**The `_sack` vs `_uk` "shared concept" columns are near-perfect duplicates** —
they were checked for agreement above:

| pair | agreement |
|---|---|
| `surface_sack` vs `surface_uk` | 100% (only differs by casing: `Hard` vs `hard`) |
| `best_of_sack` vs `best_of_uk` | 100% |
| `round_sack` vs `round_uk` / `canonical_round` | 100% |
| `winner_id` vs `canonical_winner_id`, `loser_id` vs `canonical_loser_id` | 100% |
| `winner_rank_sack` vs `winner_rank_uk` | 99.9% |
| `loser_rank_points_sack` vs `loser_rank_points_uk` | 99.7% |
| `winner_rank_points_sack` vs `winner_rank_points_uk` | 97.9% |
| `loser_rank_sack` vs `loser_rank_uk` | 97.8% |

So `canonical_*` columns are just the Sackmann identity re-exported through the
linker (100% match) — pure redundancy. Rank/points differ by ~0.1-2.2%, almost
certainly a ranking-snapshot-date difference between sources; not worth
carrying both. **Rule applied: keep the Sackmann version, drop the UK
duplicate.**

### Keep — core Sackmann (gold standard)
- **Identity/keys**: `canonical_match_key`, `tourney_id`, `tourney_name`, `tourney_date`, `tourney_level`, `draw_size`, `match_num`
- **Match context**: `surface` (←`surface_sack`), `round` (←`round_sack`), `best_of` (←`best_of_sack`), `match_status` (←`match_status_sack`)
- **Players (pre-match, safe to use as predictors)**: `winner_id/name/hand/ht/ioc/age/seed/entry`, `loser_id/name/hand/ht/ioc/age/seed/entry`, `winner_rank`, `winner_rank_points`, `loser_rank`, `loser_rank_points` (←`_sack`)
- **Outcome / in-match stats (post-match only — see leakage warning below)**: `score`, `minutes`, `sets_played`, `w_*`/`l_*` serve stats, `set{1..5}_winner/loser`, `set{1..5}_tiebreak_winner/loser`

### Keep — UK supplemental (things Sackmann doesn't have)
- **Betting odds** (main reason to keep UK data at all): `odds_b365_winner/loser`, `odds_pinnacle_winner/loser`, `odds_max_winner/loser`, `odds_avg_winner/loser`
- **Extra context**: `series` (tour-tier label, denser categories than `tourney_level`), `location`, `is_outdoor`, `players_remaining`

### Keep — linkage QA (for filtering, not modelling features)
- `match_method`, `review_flag`, `data_quality_flag` — use to drop/inspect low-confidence rows (61 `rank_mismatch` + 13 `missing_rank` out of 2702 here) before serious modelling.

### Drop — redundant duplicates of the Sackmann columns above
`surface_uk`, `round_uk`, `best_of_uk`, `winner_name_uk`, `loser_name_uk`,
`match_status_uk`, `winner_rank_uk`, `winner_rank_points_uk`, `loser_rank_uk`,
`loser_rank_points_uk`, `canonical_winner_id`, `canonical_winner_name`,
`canonical_loser_id`, `canonical_loser_name`, `canonical_tourney_id`,
`canonical_tourney_name`, `canonical_round`,
`winner_sets`, `loser_sets`, `winner_set_{1..5}_games`, `loser_set_{1..5}_games`
(redundant scoreline already captured by Sackmann's `score`/`set{i}_winner/loser`).

### Drop — pure bookkeeping (not features)
`tour_sack`, `tour_uk`, `source`, `source_year`, `year`, `match_type`,
`match_level`, `uk_tournament_id`, `tournament_name` (UK), `official_tournament_id`,
`source_event_key`, `source_match_key` — all constant, ID-only, or duplicate the
`tourney_*`/`canonical_match_key` columns already kept.

### ⚠️ Leakage warning
`minutes`, all `w_*`/`l_*` serve stats, and per-set game/tiebreak columns are
only known **after** the match is played. Fine for descriptive EDA (e.g. "how
does surface affect ace rate"), but if the eventual model predicts match
outcome from pre-match info, those must be excluded from `X` (rank, age,
odds, surface, round, height, handedness are the legitimate pre-match
features).

```python
# rename Sackmann columns back to their unsuffixed "canonical" name and
# select the recommended feature set (keep everything switch: set KEEP_ALL=True)
KEEP_ALL = False

sack_rename = {f"{concept}_sack": concept for concept in shared_concepts}

core_sack = [
    "canonical_match_key", "tourney_id", "tourney_name", "tourney_date",
    "tourney_level", "draw_size", "match_num",
    "surface", "round", "best_of", "match_status",
    "winner_id", "winner_name", "winner_hand", "winner_ht", "winner_ioc",
    "winner_age", "winner_seed", "winner_entry",
    "loser_id", "loser_name", "loser_hand", "loser_ht", "loser_ioc",
    "loser_age", "loser_seed", "loser_entry",
    "winner_rank", "winner_rank_points", "loser_rank", "loser_rank_points",
    "score", "minutes", "sets_played",
    "w_ace", "w_df", "w_svpt", "w_1stIn", "w_1stWon", "w_2ndWon", "w_SvGms",
    "w_bpSaved", "w_bpFaced",
    "l_ace", "l_df", "l_svpt", "l_1stIn", "l_1stWon", "l_2ndWon", "l_SvGms",
    "l_bpSaved", "l_bpFaced",
] + [f"set{i}_{side}" for i in range(1, 6) for side in ("winner", "loser")] \
  + [f"set{i}_tiebreak_{side}" for i in range(1, 6) for side in ("winner", "loser")]

uk_supplemental = [
    "odds_b365_winner", "odds_b365_loser",
    "odds_pinnacle_winner", "odds_pinnacle_loser",
    "odds_max_winner", "odds_max_loser",
    "odds_avg_winner", "odds_avg_loser",
    "series", "location", "is_outdoor", "players_remaining",
]

linkage_qa = ["match_method", "review_flag", "data_quality_flag"]

if KEEP_ALL:
    df_model = df_merge.rename(columns=sack_rename)
else:
    keep_cols = core_sack + uk_supplemental + linkage_qa
    df_model = df_merge.rename(columns=sack_rename)[keep_cols]

print(df_model.shape)
display(df_model.head())
```

```python
df_model.sample(n=25).to_csv("model.csv", index=False)
```

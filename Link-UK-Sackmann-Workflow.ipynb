## The Process

Yes. I think your tournament mapper lets us improve the GitHub approach substantially.

The linked script starts by joining Tennis-Data and Sackmann on (tour, winner_rank, loser_rank), then uses a fairly broad date window to keep the candidate set manageable. It drops any ambiguous one-to-many candidates, then learns player-name aliases from the accepted links.

Your version can make the first pass much tighter:


```text
Tennis-Data match
        ↓
canonical tournament ID
        ↓
only Sackmann matches in that tournament
        ↓
winner rank + loser rank
        ↓
unique canonical match
        ↓
winner_id / loser_id

```

That should be both faster and safer.

So the first job of the linker should be to attach official_tournament_id to each match dataframe.

### 5. Round should be supporting evidence

I wouldn't put round in the initial merge key.

Why?

Because you've already done enough tennis-data cleaning to know that source fields can occasionally be imperfect.

Instead:

```text
Tournament       HARD BLOCK
Winner rank      PRIMARY MATCH SIGNAL
Loser rank       PRIMARY MATCH SIGNAL
Round            VALIDATION / DISAMBIGUATION
Names            VALIDATION / FALLBACK
Rank points      VALIDATION / FALLBACK
```

If both datasets have normalized rounds like:

```text
R128
R64
R32
R16
QF
SF
F
RR
```

then after generating candidates:

````python
# classify_rank_candidates() already computed this above - shown here for reference.
candidates["round_agrees"].value_counts(dropna=False)
````

Most unique rank matches should agree.

If they don't, that's something worth investigating rather than automatically rejecting.

### 6. Rank points are another surprisingly good signal

You have:

UK:
```text
winner_rank
loser_rank
winner_rank_points
loser_rank_points
```

and Sackmann also contains:

```text
winner_rank
loser_rank
winner_rank_points
loser_rank_points
```

So after exact rank matching, we can validate:

````python
# classify_rank_candidates() already computed these too, named
# winner_rank_points_diff/loser_rank_points_diff (not winner_points_diff/loser_points_diff).
candidates[["winner_rank_points_diff", "loser_rank_points_diff"]].describe()
````

````python
REVIEW_COLUMNS = [
    # identity / tournament context
    "source_match_key",
    "canonical_match_key",
    "official_tournament_id",
    "tournament_name",
    "tourney_name",
    "location",
    "match_date",
    "tourney_date",
    # the rank pair the match was blocked/joined on
    "winner_rank",
    "loser_rank",
    # names, to eyeball whether this is even the same match
    "winner_name_uk",
    "winner_name_sk",
    "loser_name_uk",
    "loser_name_sk",
    # supporting evidence that flagged this row as suspicious
    "round_uk",
    "round_sk",
    "round_agrees",
    "winner_rank_points_uk",
    "winner_rank_points_sk",
    "winner_rank_points_diff",
    "loser_rank_points_uk",
    "loser_rank_points_sk",
    "loser_rank_points_diff",
    "is_unique",
]


sorted_rnk_point_diff = (
    (candidates[["winner_rank_points_diff", "loser_rank_points_diff"]].sum(axis=1)) ** 2
).sort_values(ascending=False)

sorted_rnk_point_diff = sorted_rnk_point_diff.loc[sorted_rnk_point_diff > 0]
top_idx = sorted_rnk_point_diff.index[0]

print(sorted_rnk_point_diff)

top_candidate = candidates.loc[top_idx, REVIEW_COLUMNS]

display(top_candidate)
````

I wouldn't require equality because the sources could use slightly different ranking snapshots.

But if we see:

```text
rank          exact
rank points   exact/nearly exact
round         exact
tournament    exact
```

that's extraordinarily strong evidence.

### 9. Where the GitHub name-map idea comes next

This is the part I absolutely would borrow from the reference implementation.

It learns:

```text
Tennis-Data name
        ↓
Sackmann player ID
```


only from matches that have already been independently linked by rankings. It requires repeated observations and 90% agreement before trusting the alias.

That's clever because we're not saying:

```text
"Sinner J." looks like "Jannik Sinner"

```

We're saying:

```text
Across 47 matches independently identified through
tournament + ranks:

"Sinner J." corresponded to player 206173
47 / 47 times.

```

That is a much stronger entity-resolution mechanism.

And ours is even safer because our seed matches have already been tournament-blocked.

### 10. The algorithm I see emerging

I'd structure the eventual resolver as:

```text
PASS 0 — Tournament resolution
─────────────────────────────
UK source_event_key
         ↓
official_tournament_id
         ↑
Sackmann tourney_id


PASS 1 — Exact rank linkage
───────────────────────────
official tournament
+ winner rank
+ loser rank
         ↓
unique 1:1 match


PASS 2 — Learn player aliases
─────────────────────────────
trusted Pass-1 matches
         ↓
UK name → canonical player ID
         ↓
require repeated dominant mapping


PASS 3 — Player-pair linkage
────────────────────────────
official tournament
+ winner player ID
+ loser player ID
         ↓
unique canonical match


PASS 4 — Residual scoring
─────────────────────────
tournament
+ rank proximity
+ player-name compatibility
+ round
+ rank points
+ opponent identity
         ↓
high confidence / review / unmatched
```

And importantly:


```text
Never fuzzy match the entire annual dataset.
```

Everything after Pass 0 happens inside one canonical tournament.

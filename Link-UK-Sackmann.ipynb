## The Process

Yes. I think your tournament mapper lets us improve the GitHub approach substantially.

The linked script starts by joining Tennis-Data and Sackmann on (tour, winner_rank, loser_rank), then uses a fairly broad date window to keep the candidate set manageable. It drops any ambiguous one-to-many candidates, then learns player-name aliases from the accepted links.

Your version can make the first pass much tighter:


```text
Tennis-Data match
        ↓
canonical tournament ID
        ↓
only Sackmann matches in that tournament
        ↓
winner rank + loser rank
        ↓
unique canonical match
        ↓
winner_id / loser_id

```

That should be both faster and safer.

### 1. First enrich both datasets with the same tournament ID

Your tournament mapper is really a bridge table:

```plaintext
year  source          source_tournament_id          official_tournament_id
2025  tennis_data_uk  2025_29_paris_french_open    520
2025  sackmann        2025-520                      520
```

So the first job of the linker should be to attach official_tournament_id to each match dataframe.

I would write a general helper:

````python
import pandas as pd


def attach_tournament_ids(
    matches: pd.DataFrame,
    tournament_mapper: pd.DataFrame,
    *,
    source: str,
    match_tournament_col: str,
) -> pd.DataFrame:
    """Attach canonical tournament IDs to a source's match dataframe.

    If `matches` has no `year` column (e.g. Sackmann's raw match data), it's
    derived from the year prefix embedded in `tourney_id` ("{year}-{number}").
    """
    matches = matches.copy()
    if "year" not in matches.columns:
        matches["year"] = matches["tourney_id"].str.split("-").str[0].astype(int)

    mapper = (
        tournament_mapper.loc[
            tournament_mapper["source"].eq(source),
            [
                "year",
                "source_tournament_id",
                "official_tournament_id",
            ],
        ]
        .drop_duplicates()
    )

    result = matches.merge(
        mapper,
        left_on=["year", match_tournament_col],
        right_on=["year", "source_tournament_id"],
        how="left",
        validate="many_to_one",
    )

    return result.drop(columns="source_tournament_id")
````

````python
uk = attach_tournament_ids(
    matches=df_uk,
    tournament_mapper=df_tournament,
    source="tennis_data_uk",
    match_tournament_col="source_event_key",
)

sackmann = attach_tournament_ids(
    matches=df_sackmann,
    tournament_mapper=df_tournament,
    source="sackmann",
    match_tournament_col="tourney_id",
)
````

### 2. Validate tournament coverage immediately

Before touching match linkage:

````python
def tournament_linkage_summary(df: pd.DataFrame) -> pd.Series:
    return pd.Series(
        {
            "matches": len(df),
            "mapped": df["official_tournament_id"].notna().sum(),
            "unmapped": df["official_tournament_id"].isna().sum(),
            "coverage": df["official_tournament_id"].notna().mean(),
        }
    )
````

````python
print("------UK Check------")
print(tournament_linkage_summary(uk))
print()
print("------Sackmann Check------")
print(tournament_linkage_summary(sackmann))
````

### 3. Bootstrap links using exact ranking pairs

Once tournaments are aligned, the first match resolver can be surprisingly simple.

For Tennis-Data:

```plaintext
winner_rank
loser_rank
```

For Sackmann:

```plaintext
winner_rank
loser_rank
```

So candidate generation becomes an indexed dataframe merge:

````python
def generate_rank_candidates(
    uk: pd.DataFrame,
    sackmann: pd.DataFrame,
) -> pd.DataFrame:
    """Generate canonical match candidates using tournament + rank pairs."""

    uk_ranked = uk.dropna(
        subset=[
            "official_tournament_id",
            "winner_rank",
            "loser_rank",
        ]
    ).copy()

    sk_ranked = sackmann.dropna(
        subset=[
            "official_tournament_id",
            "winner_rank",
            "loser_rank",
        ]
    ).copy()

    for df in (uk_ranked, sk_ranked):
        df["winner_rank"] = df["winner_rank"].astype("Int64")
        df["loser_rank"] = df["loser_rank"].astype("Int64")

    candidates = uk_ranked.merge(
        sk_ranked,
        on=[
            "year",
            "tour",
            "official_tournament_id",
            "winner_rank",
            "loser_rank",
        ],
        how="inner",
        suffixes=("_uk", "_sk"),
    )

    return candidates
````

That's the core of the algorithm.

Notice what we've eliminated.

The GitHub implementation uses:

```text
tour
winner_rank
loser_rank
± date window
```

because it doesn't know that the two rows belong to the same tournament.

We get:

```text
year
tour
official_tournament_id
winner_rank
loser_rank
```

which is substantially stronger.

````python
candidates = generate_rank_candidates(uk, sackmann)
print("------Candidates Check------")
print(tournament_linkage_summary(candidates))
````

### 4. Don't immediately accept every candidate

This part of the original script is especially worth keeping:

ambiguous matches are dropped rather than guessed.

But I would make the ambiguity visible rather than simply discard it.

First give Sackmann a stable match key:

````python
candidates["canonical_match_key"] = (
    candidates["tourney_id"].astype(str)
    + "_"
    + candidates["match_num"].astype(str)
)
````

````python
def classify_rank_candidates(
    candidates: pd.DataFrame,
) -> pd.DataFrame:
    candidates = candidates.copy()

    candidates["source_candidate_count"] = (
        candidates.groupby("source_match_key")[
            "canonical_match_key"
        ].transform("nunique")
    )

    candidates["canonical_candidate_count"] = (
        candidates.groupby("canonical_match_key")[
            "source_match_key"
        ].transform("nunique")
    )

    candidates["is_unique"] = (
        candidates["source_candidate_count"].eq(1)
        & candidates["canonical_candidate_count"].eq(1)
    )

    return candidates
````

````python
candidates = classify_rank_candidates(candidates)
print("Ambiguous candidates:", (~candidates["is_unique"]).sum())
print("Unique candidates:", candidates["is_unique"].sum())
````

Now we can distinguish:

```text
source_match       canonical candidates
----------------------------------------
match_001          1       → accept
match_002          1       → accept
match_003          2       → ambiguous
```

rather than silently throwing information away.

### 6. Rank points are another surprisingly good signal

You have:

UK:
```text

loser_rank
winner_rank_points
loser_rank_points
```

and Sackmann also contains:

```text
winner_rank
loser_rank
winner_rank_points
loser_rank_points
```

So after exact rank matching, we can validate:

````python
candidates["winner_points_diff"] = (
    candidates["winner_rank_points_uk"]
    - candidates["winner_rank_points_sk"]
).abs()

candidates["loser_points_diff"] = (
    candidates["loser_rank_points_uk"]
    - candidates["loser_rank_points_sk"]
).abs()
````

### 9. Where the GitHub name-map idea comes next

This is the part I absolutely would borrow from the reference implementation.

It learns:

```text
Tennis-Data name
        ↓
Sackmann player ID
```


only from matches that have already been independently linked by rankings. It requires repeated observations and 90% agreement before trusting the alias.

That's clever because we're not saying:

```text
"Sinner J." looks like "Jannik Sinner"

```

We're saying:

```text
Across 47 matches independently identified through
tournament + ranks:

"Sinner J." corresponded to player 206173
47 / 47 times.

```

That is a much stronger entity-resolution mechanism.

And ours is even safer because our seed matches have already been tournament-blocked.

### 10. The algorithm I see emerging

I'd structure the eventual resolver as:

```text
PASS 0 — Tournament resolution
─────────────────────────────
UK source_event_key
         ↓
official_tournament_id
         ↑
Sackmann tourney_id


PASS 1 — Exact rank linkage
───────────────────────────
official tournament
+ winner rank
+ loser rank
         ↓
unique 1:1 match


PASS 2 — Learn player aliases
─────────────────────────────
trusted Pass-1 matches
         ↓
UK name → canonical player ID
         ↓
require repeated dominant mapping


PASS 3 — Player-pair linkage
────────────────────────────
official tournament
+ winner player ID
+ loser player ID
         ↓
unique canonical match


PASS 4 — Residual scoring
─────────────────────────
tournament
+ rank proximity
+ player-name compatibility
+ round
+ rank points
+ opponent identity
         ↓
high confidence / review / unmatched
```

And importantly:


```text
Never fuzzy match the entire annual dataset.
```

Everything after Pass 0 happens inside one canonical tournament.
